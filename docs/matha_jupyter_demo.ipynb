## 1. 意图分解引擎测试

````python
from src.intent.intent_decomposer import IntentDecomposer

ide = IntentDecomposer()

# 测试不同长度的文本
test_cases = [
    ("计算 3 + 5", "短文本"),
    ("计算 100 以内所有素数，并求它们的和", "中等文本"),
    ("求解方程 x^2 - 3x + 2 = 0，然后验证 x=1 和 x=2 是否满足方程，最后计算这两个解的乘积", "长文本"),
]

for text, desc in test_cases:
    root = ide.decompose(text)
    print(f"【{desc}】")
    print(f"  输入: {text}")
    print(f"  类型: {root.intent_type.name}")
    print(f"  置信度: {root.confidence:.2f}")
    print(f"  子意图数: {len(root.children)}")
    print()
````

## 4. 算术标准库测试

````python
from src.stdlib.arithmetic import (
    sieve_of_eratosthenes, factorial, is_prime,
    gcd, lcm, combination, permutation,
    sin, cos, tan, sqrt, power
)

print("【算术标准库】")

# 素数筛
primes = sieve_of_eratosthenes(100)
print(f"  100 以内素数 ({len(primes)} 个): {primes}")

# 阶乘
print(f"  10! = {factorial(10)}")

# GCD/LCM
print(f"  gcd(12, 18) = {gcd(12, 18)}")
print(f"  lcm(4, 6) = {lcm(4, 6)}")

# 组合数学
print(f"  C(10, 3) = {combination(10, 3)}")
print(f"  P(10, 3) = {permutation(10, 3)}")

# 三角函数
print(f"  sin(π/2) = {sin(3.14159/2):.4f}")
print(f"  cos(0) = {cos(0):.4f}")
````

## 5. 代数标准库测试

````python
from src.stdlib.algebra import (
    solve_quadratic, factor_integer,
    Polynomial, poly_derivative
)

print("【代数标准库】")

# 二次方程求解
roots = solve_quadratic(1, -3, 2)
print(f"  x² - 3x + 2 = 0 的解: {roots}")

# 整数因式分解
print(f"  60 的因式分解: {factor_integer(60)}")
print(f"  100 的因式分解: {factor_integer(100)}")

# 多项式运算
p1 = Polynomial([1, 2, 1])  # 1 + 2x + x²
p2 = Polynomial([1, -1])    # 1 - x
print(f"  p1 = {p1}")
print(f"  p1 + p2 = {p1 + p2}")
print(f"  p1'(x) = {Polynomial(poly_derivative(p1.coefficients))}")
````

## 6. 微积分标准库测试

````python
import math
from src.stdlib.calculus import (
    derivative, integral, newton_method,
    taylor_series, infinite_sum
)

print("【微积分标准库】")

# 数值微分
print(f"  d/dx(x²) at x=3: {derivative(lambda x: x**2, 3):.4f} (期望 6.0)")
print(f"  d/dx(sin(x)) at x=π/2: {derivative(math.sin, math.pi/2):.4f} (期望 0.0)")

# 数值积分
print(f"  ∫[0,1] x²dx = {integral(lambda x: x**2, 0, 1):.4f} (期望 0.3333)")
print(f"  ∫[0,π] sin(x)dx = {integral(math.sin, 0, math.pi):.4f} (期望 2.0)")

# 牛顿法求根
root, iters = newton_method(lambda x: x**2 - 2, 1.0)
print(f"  √2 ≈ {root:.10f} ({iters} 次迭代)")

# 泰勒展开
print(f"  e^1 ≈ {taylor_series(math.exp, 0, 1, 20):.6f} (期望 2.71828)")

# 无穷级数
s, n = infinite_sum(lambda k: 1 / (2 ** k), max_terms=100)
print(f"  Σ(1/2^n) = {s:.6f} (期望 2.0, 用了 {n} 项)")
````

## 7. 逻辑标准库测试

````python
from src.stdlib.logic import (
    AND, OR, NOT, IMPLIES, IFF, XOR,
    truth_table, print_truth_table,
    set_union, set_intersection, set_difference,
    check_tautology, check_contradiction
)

print("【逻辑标准库】")

# 命题逻辑基本运算
print(f"  AND(T,F)={AND(True,False)}, OR(T,F)={OR(True,False)}")
print(f"  IMPLIES(T,F)={IMPLIES(True,False)}, IFF(T,T)={IFF(True,True)}")

# 真值表
print("\n  P→Q 真值表:")
print_truth_table(2, lambda v: IMPLIES(v[0], v[1]), ["P", "Q"])

# 永真式/矛盾式
print(f"  P∨¬P 是永真式: {check_tautology(1, lambda v: OR(v[0], NOT(v[0])))}")
print(f"  P∧¬P 是矛盾式: {check_contradiction(1, lambda v: AND(v[0], NOT(v[0])))}")

# 集合运算
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}
print(f"\n  A∪B = {set_union(A, B)}")
print(f"  A∩B = {set_intersection(A, B)}")
print(f"  A\\B = {set_difference(A, B)}")
````

## 8. 完整流程演示：计算 100 以内素数

````python
from src.intent.intent_decomposer import IntentDecomposer
from src.intent.llm_parser import LLMIntentParser
from src.intent.mir_generator import MIRGenerator
from src.stdlib.arithmetic import sieve_of_eratosthenes

print("=" * 60)
print("  完整流程演示：计算 100 以内所有素数")
print("=" * 60)

# 步骤 1: 自然语言输入
text = "计算 100 以内所有素数，并求它们的和"
print(f"\n【步骤 1】自然语言输入")
print(f"  {text}")

# 步骤 2: 意图分解
print(f"\n【步骤 2】意图分解")
ide = IntentDecomposer()
root = ide.decompose(text)
print(f"  意图类型: {root.intent_type.name}")
print(f"  置信度: {root.confidence:.2f}")
print(f"  子意图数: {len(root.children)}")

# 步骤 3: LLM 解析
print(f"\n【步骤 3】LLM 意图解析")
parser = LLMIntentParser()
intent = parser.parse(text)
print(f"  解析类型: {intent.intent_type.name}")
print(f"  建议代码: {intent.suggested_code}")

# 步骤 4: MIR 代码生成
print(f"\n【步骤 4】MIR 代码生成")
generator = MIRGenerator()
mir_node = generator.generate(intent)
print(mir_node.to_math_code())

# 步骤 5: 执行计算
print(f"\n【步骤 5】执行计算")
primes = sieve_of_eratosthenes(100)
total = sum(primes)
print(f"  素数列表: {primes}")
print(f"  素数个数: {len(primes)}")
print(f"  素数之和: {total}")

print(f"\n{'=' * 60}")
print(f"  ✅ 流程完成!")
print(f"{'=' * 60}")
````

## 9. 批量处理多个意图

````python
from src.intent.intent_decomposer import IntentDecomposer
from src.intent.llm_parser import LLMIntentParser
from src.stdlib.arithmetic import sieve_of_eratosthenes, factorial, gcd
from src.stdlib.algebra import solve_quadratic
from src.stdlib.calculus import derivative, integral
import math

ide = IntentDecomposer()
parser = LLMIntentParser()

tasks = [
    ("计算 100 以内所有素数", lambda: sieve_of_eratosthenes(100)),
    ("计算 10 的阶乘", lambda: factorial(10)),
    ("求解 x²-5x+6=0", lambda: solve_quadratic(1, -5, 6)),
    ("计算 sin(x) 在 [0,π] 上的积分", lambda: integral(math.sin, 0, math.pi)),
    ("求 12 和 18 的最大公约数", lambda: gcd(12, 18)),
]

print("=" * 60)
print("  批量处理多个数学任务")
print("=" * 60)

for text, compute_fn in tasks:
    intent = parser.parse(text)
    result = compute_fn()
    print(f"\n【{text}】")
    print(f"  意图类型: {intent.intent_type.name}")
    print(f"  结果: {result}")

print(f"\n{'=' * 60}")
print(f"  共处理 {len(tasks)} 个任务")
print(f"{'=' * 60}")
````
